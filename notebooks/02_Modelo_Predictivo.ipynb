## Datos y alcance

2023: 1,003,178 episodios. 2024: 883,959. 2025: 824,978. Fuente: enunciado del caso proporcionado, sin verificación externa de estas cifras. La estimación 2026 no forma parte del entrenamiento.

Requiere numpy, matplotlib y scikit-learn disponibles en el entorno de ejecución. El modelo es académico: tres observaciones, sin validación externa ni inferencia causal.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

X = np.array([[2023], [2024], [2025]])
y = np.array([1003178, 883959, 824978])

modelo = LinearRegression()
modelo.fit(X, y)

anio_nuevo = [[2026]]
prediccion_2026 = modelo.predict(anio_nuevo)[0]
print(round(prediccion_2026))
```

```python
assert abs(prediccion_2026 - 725838.3333333334) < 0.01
print("Pendiente:", modelo.coef_[0])
print("Estimación sin redondear:", prediccion_2026)
```

```python
anios_grafico = np.array([[2023], [2024], [2025], [2026]])
plt.figure(figsize=(9, 5))
plt.plot(X.ravel(), y, "o-", color="teal", label="Observados en el caso")
plt.plot(anios_grafico.ravel(), modelo.predict(anios_grafico), "--", color="purple", label="Regresión lineal")
plt.scatter([2026], [prediccion_2026], color="purple", s=90, label="Estimación 2026", zorder=5)
plt.xticks(anios_grafico.ravel())
plt.xlabel("Año")
plt.ylabel("Episodios de IRA")
plt.title("IRA infantil en Perú · Ejercicio académico")
plt.ticklabel_format(style="plain", axis="y")
plt.grid(alpha=0.2)
plt.legend()
plt.tight_layout()
plt.show()
```

## Interpretación y limitaciones

El ajuste muestra una pendiente descendente de −89,100 episodios por año. Si continúa esa tendencia lineal, la estimación 2026 es aproximadamente 725,838 episodios. No es un pronóstico epidemiológico validado: solo se utilizan tres datos anuales y no se modela estacionalidad ni otros determinantes. No hay partición entrenamiento/prueba, intervalos de incertidumbre ni medidas inventadas de precisión.

El ajuste ocurre en `modelo.fit(X, y)` y la predicción en `modelo.predict()`. La web muestra un cálculo equivalente en JavaScript, no ejecuta este notebook.
