### 1. Datos del caso

Utilizamos los episodios de IRA indicados en el **material del examen parcial**:

| Año | Episodios de IRA |
|---:|---:|
| 2023 | 1 003 178 |
| 2024 | 883 959 |
| 2025 | 824 978 |

- **X:** año de registro (variable de entrada).
- **y:** cantidad de episodios de IRA (variable objetivo).

Los datos son los proporcionados en el caso académico; aquí no se ha verificado su procedencia con una fuente estadística externa.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

X = np.array([[2023], [2024], [2025]])
y = np.array([1003178, 883959, 824978])

print("Años:", X.ravel())
print("Episodios:", y)
```

### 2. Selección y entrenamiento del modelo

Se utiliza **`LinearRegression`** porque queremos representar, de manera sencilla, cómo cambia el número de episodios según el año. El modelo encuentra una línea que se ajusta a los datos.

La línea **`modelo.fit(X, y)` es el entrenamiento**: ahí el modelo calcula su relación lineal con los tres ejemplos históricos.

```python
modelo = LinearRegression()
modelo.fit(X, y)

print("Modelo de regresión lineal entrenado.")
```

### 3. Predicción para 2026

La instrucción **`modelo.predict(...)`** utiliza el modelo ya entrenado para estimar el valor del año 2026. Ese año **no se usó en el entrenamiento**.

```python
anio_nuevo = np.array([[2026]])
prediccion_2026 = modelo.predict(anio_nuevo)[0]

print("Predicción de episodios de IRA para 2026:")
print(round(prediccion_2026))
```

### 4. Gráfico de datos y predicción

El gráfico distingue los registros utilizados para entrenar, la línea de regresión y el punto estimado para 2026.

```python
X_grafica = np.array([[2023], [2024], [2025], [2026]])
y_grafica = modelo.predict(X_grafica)

plt.figure(figsize=(9, 5))

plt.scatter(X.ravel(), y, label="Datos del caso (2023-2025)", s=90)

plt.plot(X_grafica.ravel(), y_grafica, label="Regresión lineal")

plt.scatter([2026], [prediccion_2026], label="Predicción 2026", s=110, marker="X")

plt.title("Predicción de episodios de IRA en niños menores de 5 años")
plt.xlabel("Año")
plt.ylabel("Episodios de IRA")
plt.xticks([2023, 2024, 2025, 2026])
plt.ticklabel_format(style="plain", axis="y")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5. Resultado e interpretación

El modelo **estima aproximadamente 725 838 episodios de IRA para 2026** si se mantiene la tendencia lineal de estos tres registros.

**Limitación importante:** este es un ejercicio académico con apenas tres datos anuales. No se realiza separación entrenamiento/prueba ni validación de precisión porque el conjunto es demasiado pequeño para una evaluación fiable. La predicción **no es un pronóstico epidemiológico validado**, y la disminución observada no demuestra por sí sola qué intervención la produjo.
