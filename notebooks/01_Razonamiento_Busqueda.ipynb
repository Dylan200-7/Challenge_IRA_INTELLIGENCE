**Objetivo:** representar el problema con un espacio de estados y encontrar una ruta mediante búsqueda BFS. Este es un **modelo conceptual académico**, no una simulación clínica.

### 1. Problema

Las infecciones respiratorias agudas (IRA) y la neumonía afectan a niños menores de cinco años en el Perú. El propósito de este ejercicio es **explorar rutas de intervención** que conduzcan conceptualmente a la reducción de casos y complicaciones.

Para ello se define un estado inicial, acciones posibles y un estado objetivo. Las transiciones del grafo **no demuestran** que una intervención cause automáticamente una disminución de casos.

### 2. Espacio de estados

| Estado | Significado |
|:---:|---|
| **E0** | Muchos casos de IRA y neumonía |
| **E1** | Mayor vacunación |
| **E2** | Prevención en hogares |
| **E3** | Atención temprana |
| **E4** | Menos contagios y casos |
| **E5** | Menos complicaciones |
| **OBJETIVO** | Reducir casos y muertes por IRA y neumonía infantil |

**Operadores considerados:** aumentar la vacunación, promover la prevención en hogares y facilitar la atención temprana.

**Restricciones consideradas:** disponibilidad de recursos, acceso a centros de salud y cobertura de las intervenciones. Son restricciones conceptuales; este grafo sencillo no calcula sus costos.

**Representación de relaciones:**

```text
                  ┌── E1: Vacunación ──┐
E0: Muchos casos ─┼── E2: Prevención ──┼─→ E4: Menos casos ─→ E5: Menos complicaciones ─→ OBJETIVO
                  └── E3: Atención ────┘
```

````python
# ===========================
# 1. RAZONAMIENTO
# ===========================

# Grafo: desde cada estado se muestran los siguientes estados posibles.
ira_neumonia = {
    "E0": ["E1", "E2", "E3"],
    "E1": ["E4"],
    "E2": ["E4"],
    "E3": ["E4"],
    "E4": ["E5"],
    "E5": ["OBJETIVO"],
    "OBJETIVO": []
}

# Nombre de cada estado.
estados = {
    "E0": "Muchos casos de IRA y neumonía",
    "E1": "Mayor vacunación",
    "E2": "Prevención en hogares",
    "E3": "Atención temprana",
    "E4": "Menos contagios y casos",
    "E5": "Menos complicaciones",
    "OBJETIVO": "Reducir casos y muertes por IRA y neumonía infantil"
}

print("ESTADO INICIAL:", estados["E0"])
print("OBJETIVO:", estados["OBJETIVO"])
````

### 3. Exploración y búsqueda BFS optimizada

Se utiliza **BFS (búsqueda en anchura)** con una cola (`deque`). La mejora incorpora un conjunto `descubiertos`, que registra cada estado en el momento de agregarlo a la cola, evitando encolar el mismo estado varias veces. La lista `visitados` conserva el orden de exploración para explicar el resultado.

La búsqueda respeta el orden definido en el grafo: primero `E1`, después `E2` y finalmente `E3`. No significa que `E1` sea mejor médicamente.

````python
from collections import deque

# ===========================
# 2. BÚSQUEDA BFS OPTIMIZADA
# ===========================
def buscar_ruta(grafo, inicio, objetivo):
    cola = deque([[inicio]])
    descubiertos = {inicio}  # Evita encolar el mismo estado más de una vez.
    visitados = []          # Guarda el orden de exploración.

    while cola:
        ruta = cola.popleft()
        nodo = ruta[-1]

        if nodo == objetivo:
            return ruta, visitados

        visitados.append(nodo)

        for vecino in grafo.get(nodo, []):
            if vecino not in descubiertos:
                descubiertos.add(vecino)
                cola.append(ruta + [vecino])

    return None, visitados

ruta, visitados = buscar_ruta(ira_neumonia, "E0", "OBJETIVO")
print("Exploración BFS:", " → ".join(visitados))
````

### 4. Resultado de la búsqueda

Se muestra la ruta obtenida y el significado de cada estado.

````python
# ===========================
# 3. RESULTADOS
# ===========================
if ruta:
    print("RUTA ENCONTRADA:")
    print(" → ".join(ruta))
    print("\nEXPLICACIÓN:")
    for estado in ruta:
        print(estado, "->", estados[estado])
else:
    print("No se encontró una ruta hacia el objetivo.")
````

### 5. Interpretación

BFS encuentra la ruta **E0 → E1 → E4 → E5 → OBJETIVO**. El grafo también admite rutas de igual longitud pasando por `E2` y `E3`. La ruta mostrada es la primera que encuentra BFS según el orden de los vecinos; **no se ha demostrado que sea la mejor intervención sanitaria**.
