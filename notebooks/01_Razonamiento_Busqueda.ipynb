## Presentación del problema

El ejercicio representa conceptualmente IRA y neumonía infantil en Perú. Objetivo: reducir casos y muertes. Las conexiones propuestas no prueban causalidad ni garantizan resultados sanitarios.

## Espacio de estados

E0: muchos casos de IRA y neumonía. E1: mayor vacunación. E2: prevención en hogares. E3: atención temprana. E4: menos contagios y casos. E5: menos complicaciones. OBJETIVO: reducir casos y muertes.

Operadores: incrementar vacunación, promover prevención en hogares y facilitar atención temprana. Restricciones: recursos, acceso y cobertura.

```python
from collections import deque

grafo = {
    "E0": ["E1", "E2", "E3"],
    "E1": ["E4"], "E2": ["E4"], "E3": ["E4"],
    "E4": ["E5"], "E5": ["OBJETIVO"], "OBJETIVO": []
}

def buscar_ruta(grafo, inicio, objetivo):
    cola = deque([[inicio]])
    descubiertos = {inicio}
    visitados = []
    while cola:
        ruta = cola.popleft()
        estado = ruta[-1]
        visitados.append(estado)
        if estado == objetivo:
            return ruta, visitados
        for vecino in grafo.get(estado, []):
            if vecino not in descubiertos:
                descubiertos.add(vecino)
                cola.append(ruta + [vecino])
    return [], visitados

ruta, visitados = buscar_ruta(grafo, "E0", "OBJETIVO")
print(" -> ".join(ruta))
```

```python
print("Visitados:", visitados)
assert ruta == ["E0", "E1", "E4", "E5", "OBJETIVO"]
assert visitados == ["E0", "E1", "E2", "E3", "E4", "E5", "OBJETIVO"]
assert all(b in grafo[a] for a, b in zip(ruta, ruta[1:]))
```

## Interpretación

BFS explora por niveles con una cola FIFO. Los nodos se marcan como descubiertos al encolarlos y como visitados al extraerlos. La primera ruta esperada es E0 → E1 → E4 → E5 → OBJETIVO. Tiene cuatro aristas, igual que las rutas por E2 y E3. La selección de E1 depende del orden de vecinos y no indica una intervención sanitaria superior.
